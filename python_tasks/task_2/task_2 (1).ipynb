```python
string = input("Enter a string: ")

count = 0

for ch in string:
    if ch.isalpha() and ch.lower() not in "aeiou":
        count += 1

print(count)
```

```python
string = input("Enter a string: ")

string = string.replace(" ", "")

max_count = 0
result = ""

for ch in string:
    count = string.count(ch)

    if count > max_count:
        max_count = count
        result = ch

print(result)
```

```python
numbers = [1, 2, 3, 4, 5]
result = {}

for num in numbers:
    if num % 2 == 0:
        result[num] = "Even"
    else:
        result[num] = "Odd"

print(result)
```

```python
sentence = input("Enter a sentence: ")

words = sentence.lower().split()

most_frequent = words[0]
max_count = 0

for word in words:
    count = words.count(word)

    if count > max_count:
        max_count = count
        most_frequent = word

print(most_frequent)
```

```python
sentence = input("Enter a sentence: ")

words = sentence.split()
lower_words = sentence.lower().split()

longest = ""

for i in range(len(words)):
    if lower_words.count(lower_words[i]) == 1:
        if len(words[i]) > len(longest):
            longest = words[i]

print(longest)
```
